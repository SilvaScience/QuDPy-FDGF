# Benchmark of Example 3: backends, evaluation strategy and time-domain route

This notebook reports the cost of the fifth-order 2Q calculation of `examples/example3_chi5_2q.ipynb`. The measurements come from scripts in `validation/benchmarks/scripts/`, which write JSON files to `validation/results/data/`; this notebook reads those files. In this notebook you will:

- compare the dense and the sparse backend on the seven pathways, and the two ways of evaluating the grid (`strategy="pointwise"` and `"factorized"`);
- follow the cost with the number of grid points;
- compare the frequency-domain map with a time-sampled calculation followed by a Fourier transform, at two broadenings;
- read the estimated cost of the same map with the original QuDPy.

Timings were taken on one core of a laptop processor (BLAS limited to one thread), so they show scaling, not optimized performance. To measure again on your machine, set `rerun = True` in the next cell (a few minutes).

## 0. Imports and options

`run_script` launches one benchmark script with a single BLAS thread. It does nothing unless `rerun` is `True`. The time-domain comparison with QuDPy also needs a clone of `SilvaScience/QuDPy` (set `QUDPY_REPO` or place it next to this repository).

```python
import json
import os
import subprocess
import sys
from pathlib import Path

sys.path.insert(0, "..")

import matplotlib.pyplot as plt
import numpy as np

from models import ANALYSIS_FIGURES, DATA, results_dirs

results_dirs()
SCRIPTS = Path("scripts")
rerun = False          # True: run the scripts below


def run_script(name, *args):
    if not rerun:
        return
    env = dict(os.environ, OMP_NUM_THREADS="1", OPENBLAS_NUM_THREADS="1", MKL_NUM_THREADS="1")
    subprocess.run([sys.executable, str(SCRIPTS / name), *args], check=True, env=env)


def load(name):
    return json.loads((DATA / name).read_text())
```

## 1. Dense versus sparse backend, pointwise versus factorized evaluation

Script `bench_example3_dense_sparse.py` computes the model of the example ($D=10$, 7 pathways, $\eta=8$ meV) with both backends and both strategies on a $31\times31$ grid over the same window, including the two mode-resolved polarizations, and the factorized evaluation on the published $151\times151$ grid. The model is closed and in its eigenbasis, so its generator is diagonal and the sparse backend solves the shifted systems by division (`diagonal_solve=True`).

```python
run_script("bench_example3_dense_sparse.py")
ds = load("example3_dense_sparse.json")

print(f"D = {ds['D']}, {ds['pathways']} pathways, eta = {ds['eta']} eV")
print(f"{'run':<34}{'time (s)':>10}{'resolvent solves':>18}")
for name, run in ds["runs"].items():
    print(f"{name:<34}{run['time_s']:>10.3f}{run['resolvent_solves']:>18}")
for name, value in ds["relative_difference"].items():
    print(f"  {name}: max|dense - sparse| / max|dense| = {value:.1e}")
print("pointwise versus factorized, worst relative difference:", {k: f"{v:.1e}" for k, v in ds["strategy_difference"].items()})
assert max(ds["relative_difference"].values()) < 1e-10
assert max(ds["strategy_difference"].values()) < 1e-10
```

The two backends agree to $\sim10^{-15}$ and so do the two strategies. On the $31\times31$ grid the factorized evaluation takes 124 resolvent solves instead of 13 454 and a few hundredths of a second; the published $151\times151$ map takes 604 solves.

### Cost against the number of grid points

```python
run_script("bench_grid_size.py")
rows = [r for r in load("grid_size.json")["rows"] if r["example"] == "example3"]

print(f"{'N':>4}{'pointwise (s)':>15}{'solves':>9}{'factorized (s)':>16}{'solves':>9}")
for n in sorted({r["N"] for r in rows}):
    p = next((r for r in rows if r["N"] == n and r["strategy"] == "pointwise"), None)
    f = next((r for r in rows if r["N"] == n and r["strategy"] == "factorized"), None)
    print(f"{n:>4}{'' if p is None else format(p['time_s'], '15.3f'):>15}{'' if p is None else p['resolvent_solves']:>9}"
          f"{'' if f is None else format(f['time_s'], '16.3f'):>16}{'' if f is None else f['resolvent_solves']:>9}")

fig, ax = plt.subplots(figsize=(6.0, 3.6))
for strategy, style in (("pointwise", "o-"), ("factorized", "s-")):
    sel = [r for r in rows if r["strategy"] == strategy]
    ax.loglog([r["N"] for r in sel], [r["time_s"] for r in sel], style, label=strategy)
ax.set_xlabel("Grid points per axis $N$")
ax.set_ylabel("Time for the map (s)")
ax.legend(frameon=False)
fig.savefig(ANALYSIS_FIGURES / "example3_benchmark_grid_size.pdf", bbox_inches="tight")
plt.show()
```

## 2. Frequency domain versus time sampling

Script `bench_time_fft_ex3.py` evaluates the $151\times151$ map by sampling the fifth-order response in time and transforming it, pair by pair like the pointwise evaluation, for the smallest sampling that reaches a relative error of $10^{-2}$ and $10^{-3}$ with respect to the frequency-domain result, at $\eta=8$ meV and $\eta=2$ meV. For the original QuDPy it runs a small pilot (75 samples per axis) and extrapolates the time and the memory of the full map, instead of running it.

```python
run_script("bench_time_fft_ex3.py")
tf = load("time_fft_ex3.json")

print(f"{'eta':>7}{'target':>8}{'N per axis':>12}{'time route (s)':>16}{'frequency (s)':>15}{'QuDPy est. (s)':>16}{'QuDPy est. (GB)':>17}")
for regime, report in tf.items():
    for target, best in report["time_route"].items():
        q = report["qudpy"][target]
        print(f"{regime:>7}{target:>8}{best['N_per_axis']:>12}{best['time_s']:>16.1f}{report['reference_time_s']:>15.1f}"
              f"{q['estimated_time_s']:>16.0f}{q['estimated_peak_GB']:>17.1f}")

fig, ax = plt.subplots(figsize=(6.0, 3.6))
for regime, marker in (("8meV", "o"), ("2meV", "s")):
    report = tf[regime]
    targets = list(report["time_route"])
    ax.loglog([float(t) for t in targets], [report["time_route"][t]["N_per_axis"] for t in targets], marker + "-", label=f"$\\eta$ = {regime[:-3]} meV")
ax.set_xlabel("Target relative error")
ax.set_ylabel("Samples per time axis")
ax.invert_xaxis()
ax.legend(frameon=False)
fig.savefig(ANALYSIS_FIGURES / "example3_benchmark_time_route.pdf", bbox_inches="tight")
plt.show()
```

### A resolved map: direct, sum over states and a factorized time route

Script `bench_resolved_map.py` repeats the comparison with a time route that is factorized like the direct evaluation (the propagated vectors are sampled once per axis), on the published windows with the grid spacing of the 2Q axis, and it adds the sum over states (diagonalization of the $100\times100$ generator). Time steps up to 1.6 eV$^{-1}$ are allowed (bandpass sampling) and the transform is a matrix product or a zero-padded FFT, whichever is cheaper. Columns `time`: samples per axis / time of the cheapest route that reaches the target error.

```python
run_script("bench_resolved_map.py", "example3")
rm = load("resolved_map.json")["example3"]

print(f"{'scenario':<14}{'N':>10}{'direct (ms)':>13}{'sparse (ms)':>13}{'sum over states (ms)':>22}"
      f"{'time 1e-2':>20}{'time 1e-3':>20}{'time 1e-4':>20}")
for name, e in rm.items():
    cells = []
    for target in ("0.01", "0.001", "0.0001"):
        row = e["time_route"][target]
        cells.append("not reached" if row is None else f"{row['N_t']} / {1e3 * row['time_s']:.1f} ms")
    print(f"{name:<14}{str(e['N_omega'][0]) + 'x' + str(e['N_omega'][1]):>10}{1e3 * e['direct_dense']['time_s']:>13.1f}"
          f"{1e3 * e['direct_sparse']['time_s']:>13.1f}{1e3 * e['sum_over_states']['time_s']:>22.2f}"
          f"{cells[0]:>20}{cells[1]:>20}{cells[2]:>20}")

for e in rm.values():
    assert e["sum_over_states"]["difference_to_dense"] < 1e-12
    assert e["direct_sparse"]["difference_to_dense"] < 1e-8
```

The time route of this script needs thousands of samples per axis at $\eta=2$ meV and its cost grows as the broadening decreases (17 s to 256 s at $10^{-2}$), whereas the factorized frequency-domain map takes about 0.05 s at both broadenings. The extrapolated cost of the original QuDPy is thousands of seconds and several GB of memory, which is why the full map was not run with it. (Its `coherence2d` also fails with QuTiP 5 when a fixed delay precedes the first scan, as in this example.) Since this time route is not factorized, the fair comparison is the next one.

## Summary

- For $D=10$ the dense backend is faster than the sparse one, and they agree to $\sim10^{-15}$.
- The $151\times151$ map takes 0.22 s with the factorized evaluation (about 300 solves) where the point-by-point evaluation would need 319 214.
- At equal resolution (the published $151\times172$ grid), the direct evaluation is 2 to 15 times faster than a factorized time route at $\eta=8$ meV and 23 to 38 times faster at $\eta=2$ meV; the sum over states is faster still (15 ms), as expected for a 100-dimensional generator that can be diagonalized.
- The original QuDPy is estimated at thousands of seconds and several GB for this map.
