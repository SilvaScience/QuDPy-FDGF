# Benchmark of Example 1: backends, evaluation strategy, time-domain route and original QuDPy

This notebook reports the cost of the open two-level calculation of `examples/example1_two_level_open.ipynb`. The measurements come from scripts in `validation/benchmarks/scripts/`, which write JSON files to `validation/results/data/`; this notebook reads those files, so opening it takes no time. In this notebook you will:

- compare the dense and the sparse backend (accuracy and time);
- compare the two ways of evaluating a grid, point by point or by cutting each pathway between the two scanned axes (`strategy="pointwise"` and `"factorized"`), and follow their cost with the number of grid points;
- compare the frequency-domain calculation with a time-sampled calculation followed by a Fourier transform;
- compare with the original QuDPy.

Timings were taken on one core of a laptop processor (BLAS limited to one thread), so they show scaling, not optimized performance. To measure again on your machine, set `rerun = True` in the next cell.

## 0. Imports and options

`run_script` launches one benchmark script with a single BLAS thread. It does nothing unless `rerun` is `True`. The comparison with the original QuDPy also needs a clone of `SilvaScience/QuDPy` (set `QUDPY_REPO` or place it next to this repository).

```python
import json
import os
import subprocess
import sys
from pathlib import Path

sys.path.insert(0, "..")

import matplotlib.pyplot as plt
import numpy as np

from models import DATA, ANALYSIS_FIGURES, results_dirs

results_dirs()
SCRIPTS = Path("scripts")
rerun = False          # True: run the scripts below (about 20 minutes in total)


def run_script(name, *args):
    if not rerun:
        return
    env = dict(os.environ, OMP_NUM_THREADS="1", OPENBLAS_NUM_THREADS="1", MKL_NUM_THREADS="1")
    subprocess.run([sys.executable, str(SCRIPTS / name), *args], check=True, env=env)


def load(name):
    return json.loads((DATA / name).read_text())
```

## 1. Dense versus sparse backend, pointwise versus factorized evaluation

Script `bench_example1_dense_sparse.py` runs the example (polarization, action population and integrated fluorescence, $41\times41$ grid) with both backends and with both evaluation strategies. `strategy="pointwise"` propagates every pathway at every grid point; `strategy="factorized"` (the default here) cuts each pathway between the two scanned intervals, evaluates the two parts once per point of their own axis and contracts them, so that the three pathways and the three observables share their resolvents.

```python
run_script("bench_example1_dense_sparse.py")
ds = load("example1_dense_sparse.json")

print(f"grid {ds['grid']}, t2 = {ds['t2']}, eta = {ds['eta']}")
print(f"{'run':<28}{'time (s)':>10}{'resolvent solves':>18}")
for name, run in ds["runs"].items():
    print(f"{name:<28}{run['time_s']:>10.3f}{run['resolvent_solves']:>18}")
print(f"\n{'observable/pathway':<22}max|dense - sparse| / max|dense|")
for key, value in ds["relative_difference"].items():
    print(f"{key:<22}{value:.1e}")
print("pointwise versus factorized, worst relative difference:", {k: f"{v:.1e}" for k, v in ds["strategy_difference"].items()})
assert ds["worst"] < 1e-12
assert max(ds["strategy_difference"].values()) < 1e-12
```

The two backends agree to machine precision on every observable and pathway, and so do the two strategies. The factorized evaluation needs 164 resolvent solves instead of 10 086 (the 82 distinct resolvents of the two axes, once for the shared forward part and once for each of the three observables) and takes a few milliseconds on either backend.

### Cost against the number of grid points

Script `bench_grid_size.py` repeats the calculation on $N\times N$ grids over the same window. The pointwise cost grows as $N^2$ (two solves per pathway and grid point), the factorized one as $N$.

```python
run_script("bench_grid_size.py")
rows = [r for r in load("grid_size.json")["rows"] if r["example"] == "example1"]

print(f"{'N':>4}{'pointwise (s)':>15}{'solves':>9}{'factorized (s)':>16}{'solves':>9}{'stored vectors':>16}")
for n in sorted({r["N"] for r in rows}):
    p = next((r for r in rows if r["N"] == n and r["strategy"] == "pointwise"), None)
    f = next((r for r in rows if r["N"] == n and r["strategy"] == "factorized"), None)
    print(f"{n:>4}{'' if p is None else format(p['time_s'], '15.3f'):>15}{'' if p is None else p['resolvent_solves']:>9}"
          f"{'' if f is None else format(f['time_s'], '16.3f'):>16}{'' if f is None else f['resolvent_solves']:>9}"
          f"{'' if f is None else f['stored_vectors']:>16}")

fig, ax = plt.subplots(figsize=(6.0, 3.6))
for strategy, style in (("pointwise", "o-"), ("factorized", "s-")):
    sel = [r for r in rows if r["strategy"] == strategy]
    ax.loglog([r["N"] for r in sel], [r["time_s"] for r in sel], style, label=strategy)
ax.set_xlabel("Grid points per axis $N$")
ax.set_ylabel("Time for the map (s)")
ax.legend(frameon=False)
fig.savefig(ANALYSIS_FIGURES / "example1_benchmark_grid_size.pdf", bbox_inches="tight")
plt.show()

full = {r["strategy"]: r for r in rows if r["N"] == 41}
assert full["factorized"]["resolvent_solves"] * 40 < full["pointwise"]["resolvent_solves"]
```

## 2. Frequency domain versus time sampling

The same spectrum can be obtained by sampling the response in time ($t_1$ and $t_3$ on a grid of step $dt$ up to a length $T$) and transforming it. Script `bench_time_fft.py` finds, for three damping regimes, the cheapest $(dt, T)$ that reaches a relative error of $10^{-2}$, $10^{-3}$ and $10^{-4}$ with respect to the frequency-domain result on the $41\times41$ grid. Its time route propagates every $(t_1,t_3)$ pair, like the pointwise evaluation; the next section compares with a time route that is factorized as well.

```python
run_script("bench_time_fft.py")
tf = load("time_fft.json")

print(f"{'regime':<14}{'width (eV)':>11}{'target':>9}{'N per axis':>12}{'time (s)':>10}{'frequency (s)':>15}")
for regime, report in tf.items():
    for target, best in report["best"].items():
        print(f"{regime:<14}{report['coherence_width']:>11.3f}{target:>9}{best['N_per_axis']:>12}"
              f"{best['time_s']:>10.3f}{report['reference_time_s']:>15.3f}")

fig, ax = plt.subplots(figsize=(6.0, 3.6))
for regime, report in tf.items():
    widths = report["coherence_width"]
    targets = list(report["best"])
    ax.loglog([float(t) for t in targets], [report["best"][t]["N_per_axis"] for t in targets], "o-", label=f"{regime} ($\\Gamma$ = {widths:.3f} eV)")
ax.set_xlabel("Target relative error")
ax.set_ylabel("Samples per time axis")
ax.invert_xaxis()
ax.legend(frameon=False)
fig.savefig(ANALYSIS_FIGURES / "example1_benchmark_time_route.pdf", bbox_inches="tight")
plt.show()
```

The number of samples grows as the damping weakens (the response lasts longer) and as the target tightens. These samples are transformed pair by pair, like the pointwise evaluation. For strong damping the time route is a few milliseconds against about 10 ms for the factorized frequency-domain map; it becomes slower once the damping is weak (0.6 to 1.6 s against 2 ms for the closed system). The next section compares with a time route that is factorized as well.

### A resolved map: direct, sum over states and a factorized time route

Script `bench_resolved_map.py` compares the routes at equal resolution. The grid spacing is a quarter of the width of the lines, $\Delta\omega=2\sqrt{3}(\Gamma_2+\eta)/4$, so the number of points per axis follows from the line width (23, 166 and 578 points for the three regimes over the 1 eV window). All routes use the factorization $S_{ij}=\langle G_j|\text{mid}\,F_i\rangle$: the direct evaluation solves the shifted systems (`direct`, dense backend; `sparse`: GMRES), the sum over states diagonalizes the generator once and divides, and the time route samples $F$ and $G$ with the one-step propagator up to a length $T$ and transforms them with the trapezoidal rule, by a matrix product on the grid or by a zero-padded FFT, whichever is cheaper. Time steps beyond the Nyquist step of the carrier are allowed (bandpass sampling: the signal lies in a band of 1 eV), and the error measured includes any aliasing. Columns `time`: samples per axis / time of the cheapest route that reaches the target error.

```python
run_script("bench_resolved_map.py", "example1")
rm = load("resolved_map.json")["example1"]

print(f"{'scenario':<14}{'N':>10}{'direct (ms)':>13}{'sparse (ms)':>13}{'sum over states (ms)':>22}"
      f"{'time 1e-2':>20}{'time 1e-3':>20}{'time 1e-4':>20}")
for name, e in rm.items():
    cells = []
    for target in ("0.01", "0.001", "0.0001"):
        row = e["time_route"][target]
        cells.append("not reached" if row is None else f"{row['N_t']} / {1e3 * row['time_s']:.1f} ms")
    print(f"{name:<14}{str(e['N_omega'][0]) + 'x' + str(e['N_omega'][1]):>10}{1e3 * e['direct_dense']['time_s']:>13.1f}"
          f"{1e3 * e['direct_sparse']['time_s']:>13.1f}{1e3 * e['sum_over_states']['time_s']:>22.2f}"
          f"{cells[0]:>20}{cells[1]:>20}{cells[2]:>20}")

for e in rm.values():
    assert e["sum_over_states"]["difference_to_dense"] < 1e-12
    assert e["direct_sparse"]["difference_to_dense"] < 1e-8
```

## 3. Original QuDPy

Script `bench_qudpy_series.py` runs `System.coherence2d` of the original QuDPy (QuTiP `mesolve`) on the same model and grid, for the $(dt,T)$ that reach each target, and compares the time with QuDPy-FDGF. The rows `tight` repeat the last two cases with tighter `mesolve` tolerances.

```python
run_script("bench_qudpy_series.py")
qd = load("qudpy_original.json")

print(f"{'case':<10}{'tight':<7}{'dt':>6}{'T':>6}{'N':>6}{'error':>9}{'QuDPy (s)':>11}{'FDGF (s)':>10}{'ratio':>7}{'peak MB':>9}")
for row in qd:
    print(f"{row['case']:<10}{str(row.get('tight', False)):<7}{row['dt']:>6.2f}{row['T']:>6.0f}{row['N_per_axis']:>6}"
          f"{row['error']:>9.1e}{row['time_qudpy_s']:>11.2f}{row['time_fdgf_s']:>10.3f}"
          f"{row['time_qudpy_s'] / row['time_fdgf_s']:>7.0f}{row['peak_MB_qudpy']:>9.0f}")

ratios = [r["time_qudpy_s"] / r["time_fdgf_s"] for r in qd]
print(f"QuDPy-FDGF is {min(ratios):.0f} to {max(ratios):.0f} times faster on these cases")
```

With the default tolerances of `mesolve`, the finest cases stay at an error of $5\times10^{-4}$ (Example 1) and $6\times10^{-3}$ (weak damping) instead of the targeted $10^{-4}$ and $10^{-3}$: the error is dominated by the integrator, not by the sampling. The `tight` rows reach $9\times10^{-5}$ and $2\times10^{-4}$, at the same cost. Memory grows as $N^2$ since all propagated states are kept. The ratios of 136 to 13 838 compare with a code that integrates every $(t_1,t_3)$ pair separately on a $41\times41$ grid that does not resolve the lines of the weak-damping case; the resolved maps of section 2 are the fairer measure of the frequency-domain advantage.

## Summary

- The two backends and the two evaluation strategies agree to machine precision; the factorized evaluation takes 164 solves instead of 10 086 for the $41\times41$ map.
- At equal resolution, the direct evaluation is comparable to a factorized time route at a relative error of $10^{-2}$ for strongly damped lines and 8 to 150 times faster at $10^{-3}$ to $10^{-4}$ (9 to 43 times for the published rates); the sum over states, which for two levels is a closed formula, is the fastest of all. The number of points of a resolved map grows as the lines narrow, and so does the cost of the direct evaluation (0.8, 5 and 36 ms for 23, 166 and 578 points per axis).
- The original QuDPy takes from 1 s to 1.5 minutes on the $41\times41$ map that QuDPy-FDGF evaluates in 6 to 8 ms.
